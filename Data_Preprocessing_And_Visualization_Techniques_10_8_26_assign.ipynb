# What does the the activation funtion do ?


An **activation function** is a non-linear transformation that is applied to the output of each neuron in a neural network. It determines whether a neuron should be activated and to what extent, based on the input it receives.

# What all are the different types of activation functions ?

 ### 1. Sigmoid funtion

*   Squashes input values to a range between 0 and 1, making it useful for binary classification problems (where the output can be interpreted as a probability).

*   Historically popular but suffers from the **vanishing gradient problem** for very large or very small inputs, which can slow down or halt learning in deep networks.

*   Output is not zero-centered, which can complicate gradient updates.

* Similar to sigmoid but output are centre by zero which often fastest convergence output during training.

   * most popular function for hidden layer in deep learning

   * Computing efficient helps migrate vanishing gradient problem compare to sigmoid/tanh and promoting spicific in the networks.

   * Can suffer from the "dying ReLU" problem, where neurons can become inactive if their input consistently falls below zero, leading to zero gradients that prevent further learning.

  * An attempt to address the "dying Relu" the problem that allow the small non-zero when input is negative

  * Helps neuron continue the learn even when their input are negative

### 5. ELU funtion:

  * It similar to RelU , but for negative inputs, it provides smoothy decreses rather than the abrupt zero.

  * Can lead to faster learning and better genneration than ReLu

### 6.Softmax function:

 * Used exclusively in the output for multi classfication problem

 * Converts a vector of arbitrary real values into a probability distribution, where each value represents the probability of the input belong to the particular class.  


# How does a Neural Network Learn? The Learning Process Explained

### Neural networks learn by iteratively adjusting their internal parameters (weights and biases)
### to minimize the difference between their predicted output and the actual target output.

## This process is often referred to as 'training' and involves several key steps:

# 1. Initialization:
* Before learning begins, the weights and biases of the neural network are typically initialized to small random values. This helps break symmetry and ensures that different neurons learn different features.

## 2. Forward Propagation:
* The input data is fed into the network. It passes through each layer, from the input layer, through hidden layers, to the output layer. At each neuron, a weighted sum of inputs is calculated, and an activation function



## 3. Loss Function (Cost Function):
* After forward propagation, the network's output is compared to the actual target output (ground truth). The loss function quantifies the error or discrepancy between the predicted and actual values.
* Common loss functions include Mean Squared Error (MSE) for regression and Cross-Entropy Loss for classification.A lower loss value indicates a better performing model.

# 4. Backpropagation:
* This is the core learning algorithm. Backpropagation calculates how much each weight and bias in the network contributed to the error measured by the loss function.
* It does this by computing the gradient of the loss function with respect to each parameter, working backward from the output layer to the input layer.
* The gradients indicate the direction and magnitude by which the parameters should be adjusted to reduce the loss.

## 5. Optimization (Gradient Descent and its variants):
* Once the gradients are computed via backpropagation, an optimizer is used to update the weights and biases.
* The most common optimization algorithm is Gradient Descent, which adjusts parameters In the direction opposite to the gradient, effectively moving towards the minimum of the loss function.
* The 'learning rate' is a hyperparameter that determines the size of the steps taken during each update.
* Too high a learning rate can cause overshooting, while too low can make training very slow.


## 6. Iteration (Epochs and Batches):
* Steps 2 through 5 are repeated many times. Each full pass of the training datasetthrough the network (forward and backward propagation) is called an 'epoch'.
* Often, the training data is divided into smaller 'batches', and parameter updatesare performed after processing each batch. This makes training more efficientand can help generalize better.

## 7. Evaluation:
* Throughout or after training, the network's performance is evaluated on a separate
* validation set (and finally on a test set) to ensure it generalizes well to unseen data and to prevent overfitting.



## The Feedback Loop in Neural Network Learning (Brief)

The learning process in a neural network is essentially a continuous feedback loop that aims to reduce errors and improve predictions. Here's a brief overview:

1.  **Forward Pass (Prediction)**: The neural network takes input data and processes it through its layers, generating an output (a prediction).

2.  **Error Calculation (Loss Function)**: This prediction is compared against the actual target value using a 'loss function'. The loss function quantifies how wrong the prediction was.

3.  **Error Propagation (Backpropagation)**: The calculated error is then propagated backward through the network (this is called backpropagation). During this step, the network determines how much each internal parameter (weights and biases) contributed to the error.

4.  **Parameter Adjustment (Optimization)**: An optimizer (like Gradient Descent) uses this error information to make small adjustments to the weights and biases of the network. These adjustments are made in a direction that is expected to reduce the error in the next forward pass.

This cycle of prediction, error calculation, and parameter adjustment repeats many times over the entire dataset (epochs), allowing the network to gradually learn the underlying patterns and make more accurate predictions.

## TensorFlow Operations for Neural Networks

TensorFlow provides several fundamental operations that are crucial for defining and manipulating tensors, which are the primary data structures in neural networks.

### 1. tf.constant()

*   **What it does:** Creates a tensor whose value cannot be changed. It holds a fixed value throughout the execution of the graph.

### 2. tf.zeros()

*   **What it does:** Creates a tensor with all elements set to zero. You specify its shape and optionally its data type.


### 3. tf.ones()

*   **What it does:** Creates a tensor with all elements set to one. Similar to `tf.zeros()`, you specify its shape and data type.


### 4. tf.random.*()

*   **What it does:** Creates tensors filled with random values from a specified distribution (e.g., normal distribution, uniform distribution). Crucial for stochastic processes.


### 5. tf.slice()

*   **What it does:** Extracts a sub-tensor from a larger tensor along specified dimensions.

## Different Types of Tensors in TensorFlow

In TensorFlow, a tensor is the primary data structure, representing multi-dimensional arrays of arbitrary shape. The 'type' of a tensor usually refers to its **rank** (number of dimensions) and its **data type** (e.g., integer, float).

### 1. Tensors by Rank (Number of Dimensions)

*   **Scalar (Rank 0 Tensor)**:
    *   **Description**: A single number. It has no dimensions.
    *   **Example**: `tf.constant(5)` or `tf.constant(3.14)`
    *   **Use in Neural Networks**: Represents individual values like a bias term for a single neuron, a learning rate, or a single sample's loss.

*   **Vector (Rank 1 Tensor)**:
    *   **Description**: An array of numbers. It has one dimension.
    *   **Example**: `tf.constant([1, 2, 3])` or `tf.constant([0.1, 0.2, 0.3])`
    *   **Use in Neural Networks**: Represents a single feature vector for one input sample, a list of biases for a layer, or gradients for a single set of weights.

*   **Matrix (Rank 2 Tensor)**:
    *   **Description**: A 2D array of numbers. It has two dimensions (rows and columns).
    *   **Example**: `tf.constant([[1, 2], [3, 4]])`
    *   **Use in Neural Networks**: Most commonly used for weights connecting two layers (e.g., an input layer to a hidden layer), where rows might be input features and columns might be output neurons.

*   **Rank 3 Tensor (e.g., for sequences or color images)**:
    *   **Description**: A 3D array of numbers. It has three dimensions.
    *   **Example**: A batch of grayscale images `(batch_size, height, width)` or a single color image `(height, width, channels)`.
    *   **Use in Neural Networks**: Input for Recurrent Neural Networks (RNNs) for sequences `(batch_size, sequence_length, features)` or single color images `(height, width, color_channels)` for Convolutional Neural Networks (CNNs).

*   **Rank 4 Tensor (e.g., for batches of color images)**:
    *   **Description**: A 4D array of numbers. It has four dimensions.
    *   **Example**: `(batch_size, height, width, channels)` for a batch of color images.
    *   **Use in Neural Networks**: Standard input format for CNNs when processing batches of color images.

*   **Higher-Rank Tensors**: Tensors can have even more dimensions depending on the complexity of the data (e.g., video data might be rank 5: `(batch, frames, height, width, channels)`).

### 2. Tensors by Data Type (Dtype)

TensorFlow supports various data types, crucial for numerical precision and memory efficiency:

*   **Floating-point types**: `tf.float16`, `tf.float32` (most common for model parameters and computations), `tf.float64`.
*   **Integer types**: `tf.int8`, `tf.int16`, `tf.int32`, `tf.int64` (common for indices, labels, or counting).
*   **Boolean type**: `tf.bool` (for logical operations and masks).
*   **String type**: `tf.string` (for raw strings, usually preprocessed into numerical representations before computations).
*   **Complex number types**: `tf.complex64`, `tf.complex128` (for specialized applications).

### Perceptron for AND Gate Logic

A single-layer perceptron is a fundamental type of artificial neural network that can learn to classify linearly separable data. The AND gate is a classic example of a linearly separable problem. The perceptron will learn weights and a bias to correctly classify the outputs for all possible AND gate inputs.

```python
import numpy as np

# 1. Define AND gate inputs and corresponding outputs
# Inputs: X = [[0,0], [0,1], [1,0], [1,1]]
# Outputs: y = [0, 0, 0, 1]
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

# 2. Define the Activation Function (Step Function for Perceptron)
def step_function(s):
    return 1 if s >= 0 else 0

# 3. Initialize Perceptron Parameters
# Random weights (2 inputs, so 2 weights)
w = np.random.rand(2)
# Random bias
b = np.random.rand(1)

# Learning rate
learning_rate = 0.1
# Number of epochs (training iterations)
epochs = 10

print(f"Initial Weights: {w}")
print(f"Initial Bias: {b[0]}")
print("----------------------")

# 4. Training the Perceptron
print("Starting Training...")
for epoch in range(epochs):
    print(f"Epoch {epoch + 1}/{epochs}")
    for i in range(len(X)): # Iterate through each input-output pair
        input_data = X[i]
        true_output = y[i]

        # Calculate weighted sum of inputs + bias (s = w1*x1 + w2*x2 + b)
        weighted_sum = np.dot(input_data, w) + b

        # Apply activation function to get predicted output
        predicted_output = step_function(weighted_sum)

        # Calculate error
        error = true_output - predicted_output

        # Update weights and bias based on error (Perceptron Learning Rule)
        w = w + learning_rate * error * input_data
        b = b + learning_rate * error

    # Optional: Print current weights and bias after each epoch
    # print(f"  Updated Weights: {w}, Bias: {b[0]}")

print("----------------------")
print("Training Complete!")
print(f"Final Weights: {w}")
print(f"Final Bias: {b[0]}")

# 5. Test the Trained Perceptron
print("\nTesting the trained Perceptron for AND gate:")
for i in range(len(X)):
    input_data = X[i]
    true_output = y[i]

    weighted_sum = np.dot(input_data, w) + b
    predicted_output = step_function(weighted_sum)

    print(f"Input: {input_data}, True Output: {true_output}, Predicted Output: {predicted_output}")

    if predicted_output == true_output:
        print("  -> Correct!")
    else:
        print("  -> Incorrect!")
```
