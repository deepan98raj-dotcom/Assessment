The `def` keyword in Python is used to **define** a function.

# 1.**What is a function?**

A function is a block of organized, reusable code that is used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

````python
def say_hello():
    print('hello')

say_hello()
say_hello()
say_hello()
say_hello()
````

# 2. Function Syntax

````python
def say_hello(name):
  print(f'Hello. Hi my dear Customer {name}. We have lovely. Choose your favorite.')

say_hello('deepan')
````

# 3. Calling a Function

Define a function does NOT execute it. we must call it using parenthesses ().


````python
def greet():
  print('Hello from greet function!')

greet()
````

#4.Function with One Parameter

Parameters are inputs passed into a funtion. they allow the funtion to work with different values.

````python
def greet_user(name):
  print('Hello',name)

greet_user('Alice')
greet_user('Bob')
greet_user('Deepan')
````

# 5. Function with Multiple Parameters

A function can accept more than one input.


````python
def add_numbers(a,b):
   print(a+b)

add_numbers(10,20)
add_numbers(100,200)
````

# 6. Return Statement

```return``` sends data back to the place where the function was called. This allows further use of the result.

````python
def add(a,b):
  return a+b

result = add(10,20)
print(result)
````

# 7. print() vs return

### Coding Explanation

  * ```print()``` displays output on screen
  * ```return``` gives output back to the program

````python
def using_print():
  print(100)

def using_return():
  return 100

using_print()
value = using_print()
print(value)

using_return()
value = using_return()
print(value)
````

# 8. Default Parameters

Default values are used when no arguments is provided

````python
def country (name,country= 'india'):
  print(name,'is from',country)

country('deepan')
country('alex','USA')
````

# 9. Returning Multiple values

python allows returning multiple values using commas.

````python
def calculation (a,b):
    return a+b,a-b,a*b

s,d,m = calculation(10,5)
print(s,d,m)
````

python lists are slow when data is very large.NumPy makes calculations faster and easier

# 3. Creating Numpy Arrays

A NumPy array stores multiple numbers in one variable.


````python
arr = np.array([1,2,3,4,5])
print(arr)
````

#4. Basic Math Operations

Numpy allows us to do math on all values at once.


````python
arr = np.array([10, 20, 30])
print(arr + 2)
print(arr - 2)
print(arr * 2)
print(arr / 2)
````

Axis controls row-wise or column-wise operations.

````python
arr = np.array([[1,2,3],[4,5,6]])
print(np.sum(arr,axis=0))
print(np.sum(arr,axis=1))
````

# 21. Copy vs view

Memory behavior of arrays.


````python
Aa = np.array([1,2,3])
Bb = a.view()
Cc = a.copy()
Aa[0] = 100
print(Aa)
print(Aa,Bb,Cc)
````

# 13. Stack & split

Combine and divide arrays.



````python
a = np.array([1,2,10])
b = np.array([4,5,6])
print(np.concatenate((a,b)))
print(np.stack((a,b)))
print(np.hstack((a,b)))
print(np.vstack((a,b)))
print(np.dstack((a,b)))
print(np.array_split(a,3))
````
