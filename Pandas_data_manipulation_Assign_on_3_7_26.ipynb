```python
#34

import pandas as pd
data = {
    'alcohol' : [8.5,9.2,10.1,11.0,9.8,10.5,11.2,8.9],
    'sugar'   : [6.2,4.8,3.5,2.1,5.0,3.0,1.8,6.5],
    'pH'      : [3.2,3.1,3.0,3.3,3.2,3.1,3.4,3.0],
    'quality' : [5,6,7,8,6,7,8,5],
    'price_category' :['Low','medium','high','high','medium','Low','medium','high']
}

df = pd.DataFrame(data)
df

```

```python
# 1. groupby() Basics
#   - Used to group data based on a column and perform calculations on each group.

# Average alcohol content by quality
df.groupby('quality')['alcohol'].mean()
```

```python
# 2.Single Aggregation

# Here we calculate mean sugur level for each price category.

df.groupby('price_category')['sugar'].mean()
```

```python
# 3. Multiple Aggregations

# - We can calculate more than one statistic at once.

df.groupby('quality')['alcohol'].agg(['mean','max','min'])
```

```python
# 4 Using agg() with Multiple Columns & Functions

df.groupby('price_category').agg({'alcohol': ['mean','max'],
                                  'sugar': 'mean',
                                  'quality': 'count'})
```

```python
# reset_index()

grouped = df.groupby('quality')['alcohol'].mean().reset_index()
grouped
```

```python
# 6. Multiple - Column GroupBy

df.groupby(['price_category','quality'])['alcohol'].mean().reset_index()
```
