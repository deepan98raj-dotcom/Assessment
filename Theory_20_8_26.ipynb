## What are the activation functions involved in CNN ?

After Convolutional Layers:
 * An activation function is almost always applied immediately after each convolutional layer. This is crucial because convolutional operations themselves are linear. Applying an activation function makes the features extracted by the filters non-linear, allowing the network to learn more complex patterns in the image data.
   * ReLU and its variants (Leaky ReLU, ELU, PReLU) are overwhelmingly the most common choices for hidden layers in CNNs. Their benefits (computational efficiency, solving vanishing gradients) are particularly valuable in deep CNNs with many convolutional layers.
   * The goal here is to transform the feature maps produced by the convolution into non-linear representations.
  * After Fully Connected Layers: CNNs typically end with one or more fully connected layers, similar to an ANN. Here, the activation functions follow the same principles as in ANNs:
    * Hidden Fully Connected Layers: ReLU or its variants are standard.
    * Output Layer: Sigmoid for binary classification, Softmax for multi-class classification, or linear for regression, depending on the task.

## What is Batch Size and Why do We Use It in NN Models?

**Batch Size** refers to the number of training examples utilized in one iteration. In simpler terms, instead of feeding the entire dataset to the neural network at once (which can be computationally expensive or even impossible for large datasets), we divide the dataset into smaller subsets called batches.

### Why do we use it?

1.  **Computational Efficiency**: Processing the entire dataset can be slow and memory-intensive. Using batches allows for more efficient computation, as it fits into memory and allows for parallel processing.

2.  **Smoother Gradient Descent**: While a batch size of 1 (Stochastic Gradient Descent) can lead to very noisy gradients and slow convergence, a very large batch size (Batch Gradient Descent) can cause the model to get stuck in local minima and generalize poorly. A moderate batch size (Mini-Batch Gradient Descent) strikes a balance, providing a more stable estimate of the gradient than SGD, and better generalization than BGD.

3.  **Regularization Effect**: Smaller batch sizes introduce more noise into the gradient updates, which can act as a form of regularization, helping the model generalize better to unseen data and preventing overfitting.

4.  **Hardware Optimization**: Modern hardware (like GPUs) is optimized for parallel computations. Processing data in batches allows for better utilization of these resources.

Choosing the right batch size is a hyperparameter tuning task and often depends on the specific dataset, model architecture, and available computational resources.

## What is a Recurrent Neural Network (RNN)?

A **Recurrent Neural Network (RNN)** is a type of artificial neural network designed to recognize patterns in sequences of data, such as text, speech, or time series data. Unlike traditional feedforward neural networks (like ANNs or CNNs) where information flows only in one direction, RNNs have 'loops' or 'recurrent connections' that allow information to persist from one step of the sequence to the next.

### Key Characteristics:

*   **Memory**: RNNs have an internal 'memory' that allows them to process sequences by taking into account previous elements in the sequence. This means the output at any given step depends not only on the current input but also on the previous computations.
*   **Sequential Data**: They are particularly well-suited for tasks involving sequential data where the order of information matters.
*   **Shared Weights**: The same set of weights is applied to each step of the sequence, making them efficient for handling variable-length sequences.

## Explain the formula used in RNN

A Recurrent Neural Network (RNN) fundamentally relies on two main equations to process sequential data, allowing it to maintain an internal 'memory' of past inputs. These equations describe how the hidden state is updated and how an output is generated at each time step.

Let's consider a simple RNN at a given time step `t`:

### 1. Hidden State Update Equation:

The hidden state `h_t` at time `t` is computed based on the previous hidden state `h_{t-1}` and the current input `x_t`. This is where the 'memory' aspect of RNNs comes from, as the network considers information from previous steps.

$$h_t = \text{tanh}(W_{hh} h_{t-1} + W_{xh} x_t + b_h)$$

Where:
*   `h_t`: The hidden state (or activation) at the current time step `t`.
*   `h_{t-1}`: The hidden state from the previous time step `t-1`.
*   `x_t`: The input at the current time step `t`.
*   `W_{hh}`: The weight matrix for the recurrent connection (hidden-to-hidden).
*   `W_{xh}`: The weight matrix for the input-to-hidden connection.
*   `b_h`: The bias vector for the hidden state.
*   `tanh`: An activation function (e.g., hyperbolic tangent, ReLU, etc.). It introduces non-linearity, allowing the network to learn complex patterns.

### 2. Output Equation:

The output `y_t` at time `t` is typically generated from the current hidden state `h_t`.

$$y_t = W_{hy} h_t + b_y$$

Where:
*   `y_t`: The output at the current time step `t`.
*   `W_{hy}`: The weight matrix for the hidden-to-output connection.
*   `b_y`: The bias vector for the output.

In some RNN architectures, another activation function (e.g., softmax for classification) might be applied to `y_t` depending on the task. These two equations form the core computational steps that allow RNNs to process sequences by iteratively updating their internal state based on new inputs and generating outputs.

## How different RNN is from CNN?

Both Recurrent Neural Networks (RNNs) and Convolutional Neural Networks (CNNs) are types of deep learning models, but they are fundamentally designed to process different types of data and excel at different tasks.

If your data has important spatial features that need to be extracted, a CNN is likely the choice. If the order and context within a sequence are critical, an RNN is usually more appropriate.
