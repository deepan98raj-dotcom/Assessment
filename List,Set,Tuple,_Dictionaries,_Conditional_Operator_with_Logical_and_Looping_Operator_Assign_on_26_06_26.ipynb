````python
num = 9
if num % 2 == 0:
  print('Even number')
else:
  print('Odd number')
````

## 3. if - elif - else statement

````python
score = 82
if score >= 90:
  print('Grade A')
elif score >= 75:
  print('Grade B')
elif score >= 90:
  print('Grade C')
else:
  print('Fail')
````

## 4. Nested if Statements

````python
age = 19
id_available = False
if age >= 18:
  if id_available:
    print('Access Granted')
  else:
    print('ID required')
else:
  print('Not allowed')
````

# Looping statements(PYTHON)

Repeating task No of times --> Until Condition satisfied


```
    For loop           |           While loop
_______________________|____________________________
                       |
know the value of n    |    value of n is not known
                       |
So keeping happens n   |    So looping happens Until  
no of times            |     condition is Satisfied

```
# What Are Loops ?
 A loop allows you to repeat actions without writing code again and again.

 Example :
```
Repeat 3 times:
  print('hello')
```
python does this for loops and while loops

# For loops

A for loop repeats over each item in a sequence (list, string,range of numbers).

```
for variable in swequence:
  # code to repeat
  ```

  

````python
for num in range(1,6):
  print(num)
````

````python
fruits = ['apple','banana','cherry']
for fruit in fruits:
  print(fruit)
````

````python
text ='python'

for ch in text:
  print(ch)
````

````python
for n in range (0,20,4):
  print(n)
````

````python
for i in range(1,10):
  if i ==  7:
   break
  print(i)
````

````python
for i in range(1,11):
  if i % 2 == 0:
    continue
  print(i)
````

# 2. While loops

A while loop repeats as long as as a condition is true



````python
i = 1
while i <= 5:
  print(i)
  i += 1
````

````python
# Example 2: Repeat until user enters correct password

password = 'python'
guess = ''

while guess != password.lower():
  guess = input('enter password')

print('password correct')
````

````python
# Using break & continue in while loop

i = 1
while i < 15:
  if i == 9:
      break
  print(i)
  i+=1
````

# 3.Nested loop(loop inside a loop)

A nested loop is a loop inside anthor loop

## Why use nested loops?

 * Creating patterns
 * comparing items
 * working with grids(rows x columns)

 ```
 for i in range(...):
    for j in range(...):
    # code
```


````python
for i in range (1,4):
  for j in range(1,4):
    print(i,j)
````

````python
# Simple star pattern

for i in range(5):
  for j in range(i + 2):
    print('*',end='')
  print()
````
